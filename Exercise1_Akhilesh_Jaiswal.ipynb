**Objective:** Create a DataFrame with random numerical columns `f1`–`f4`, plot `f2` against `f3`, calculate each row's mean in `f5`, and export all five columns to Excel.

**Concepts:** A DataFrame is a table. A scatter plot shows the relationship between two numerical variables. `axis=1` calculates across columns for each row.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_DIR = Path.cwd()
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)
np.random.seed(42)  # Reproducible random values

df = pd.DataFrame({
    "f1": np.random.randint(1, 101, 100),
    "f2": np.random.randint(1, 101, 100),
    "f3": np.random.randint(1, 101, 100),
    "f4": np.random.randint(1, 101, 100)
})
display(df.head())
print("Rows and columns:", df.shape)
print(df.dtypes)
display(df.describe())
```

## Task 1.2 — Scatter plot of f2 against f3
Each dot is one row. The x-axis is f2 and the y-axis is f3.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["f2"], df["f3"], alpha=0.7)
plt.xlabel("Column f2")
plt.ylabel("Column f3")
plt.title("Scatter Plot: Relationship Between f2 and f3")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "Exercise1_f2_f3_scatter.png", dpi=150)
plt.show()
```

## Task 1.3 — Row-wise arithmetic mean
`axis=1` means calculate the average across f1, f2, f3 and f4 for each row.

```python
df["f5"] = df[["f1", "f2", "f3", "f4"]].mean(axis=1)
display(df.head())
manual_mean = sum(df.loc[0, ["f1","f2","f3","f4"]]) / 4
print("First row f5:", df.loc[0, "f5"])
print("Manual check:", manual_mean)
```

## Task 1.4 — Export to Excel

```python
excel_path = OUTPUT_DIR / "Exercise1_Output.xlsx"
df[["f1","f2","f3","f4","f5"]].to_excel(excel_path, index=False, sheet_name="Random_Data")
print("Saved:", excel_path.resolve())
```
