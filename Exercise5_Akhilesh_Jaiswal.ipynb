## Task 5.2 — Define the chatbot function

```python
import gradio as gr

def respond(message, history):
    text = message.lower().strip()
    if any(word in text for word in ["hello", "hi", "namaste"]):
        return "Hello! Welcome to the GOT-A AI Learning Assistant. Ask me about Python, Machine Learning, Generative AI or RAG."
    if "machine learning" in text or text == "ml":
        return "Machine Learning is a branch of AI in which a computer learns patterns from examples. For example, it can learn from labelled complaints and predict the department for a new complaint."
    if "generative ai" in text or "genai" in text:
        return "Generative AI creates new content such as text, images, audio or video. A text example is drafting an acknowledgement from a citizen grievance."
    if "rag" in text:
        return "RAG means Retrieval-Augmented Generation. The system retrieves relevant information from a trusted document and supplies it as context to a language model."
    if "python" in text:
        return "Python is a programming language used with libraries such as Pandas for data analysis, OpenCV for images and Gradio for web interfaces."
    return "I am the GOT-A learning demo. Ask me about Python, Machine Learning, Generative AI or RAG."

print("Chatbot function is ready.")
```

## Task 5.3 — Build and launch the Gradio interface

```python
demo = gr.ChatInterface(
    fn=respond,
    title="GOT-A AI Learning Assistant",
    description="Welcome! Ask about Python, Machine Learning, Generative AI or RAG. This is an educational rule-based prototype.",
    examples=[
        "What is Machine Learning?",
        "Explain Generative AI with an example.",
        "What is RAG?",
        "How is Python used in AI?"
    ]
)
demo.launch(share=False)  # Local access only; do not expose private data.
```
